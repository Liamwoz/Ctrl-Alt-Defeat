This notebook compares the baseline, Random Forest, and XGBoost results after their standalone training notebooks have been run.

```python
import pandas as pd
from pathlib import Path

metric_files = [
    Path("random_forest_metrics.csv"),
    Path("xgboost_metrics.csv"),
]
comparison = pd.concat([pd.read_csv(path) for path in metric_files], ignore_index=True)
comparison.sort_values(["balanced_accuracy", "poisonous_recall"], ascending=False).reset_index(drop=True)
```

## Conclusion

Use balanced accuracy, poisonous recall, poisonous F1, and ROC AUC together when selecting the final model. Accuracy alone can hide unsafe poisonous-mushroom false negatives. The model with the strongest validation profile should be selected only after confirming that its holdout metrics and saved artifact come from the same reproducible split.
